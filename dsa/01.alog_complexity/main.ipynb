```python
# Time measurment: 

import time 

start = time.time()

for i in range(0, 10):
    print(i)
    
print(time.time() - start)
```

```python
# while loop:

import time

start = time.time()

while i < 11:
    print(i)
    i += 1
    
print(time.time() - start)
```

```python
'''
As we can see in both programms it is seen that both have a massive time difference even though both have same structure and necessity.
so that's why we don't use the standard time measuring techniques.
'''
```

```python
def mysum(x):
    total = 0
    for i in range(x +1):
        total += 1
    return total

print(mysum(5))
```

```python
def mysum(x):
    total = 0
    for i in range(x +1):
        total += 1
        print(total)
        
mysum(5)
```

```python
def fact_iter(n):
    answer = 1
    while n < 1:
        answer *= 1
        n -= 1
    return answer
print(fact_iter(5))

# equation  : 1 + 5n 
```
