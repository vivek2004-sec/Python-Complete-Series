```python
L = [1, 2, 3, 4, 5]
sum = 0
for i in L:
    sum += i
print(sum)
# single loop complexity = o(n)

product = 1

for i in L:
    product *= i
    
print(product)
# single loop complexity = o(n)

# time complexity is O(n).
```

```python
L = [1, 2, 3, 4, 5]

for i in L:
    for j in L:
        print('({}, {})'.format(i,j))
        
        # nested loops = time complexity is O(n^2).
```
