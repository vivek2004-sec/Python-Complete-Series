```python
import numpy as np
```

```python
arr1 = np.array([[1, 7], [2, 1], [3, 2]])
print(arr1)
```

```python
arr2 = np.array([[1, 1], [1, -1], [1, -2]])
print(arr2)
```

```python
array1 = np.add(arr1, arr2)
print(array1)
```

```python
array2 = np.subtract(arr1, arr2)
print(array2)
```

```python
array3 = np.multiply(arr1, arr2)
print(array3)
```

```python
a = np.array([[1, 2, 3], [1, -2, 3]])
print(a)
```

```python
c = np.dot(arr1, a)
print(c)
```

```python
det = np.linalg.det(c)
print(det)
inverse = np.linalg.inv(c)
print(inverse)
```

```python
m = np.array([[1, -1], [2, 3]])
n = np.linalg.inv(m)
print(n)
```

```python
print(np.dot(m, n))
```
