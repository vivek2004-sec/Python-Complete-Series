```python
import numpy as np
```

```python
# defining a vector
v1 = np.array([1, -1, 2])
v2 = np.array([ 2, 5, 2])

print(v1 + v2)
```

```python
print(v1 - v2)
```

```python
# scalar Multiplication 

print(3 * v1)
```

```python
print(np.linalg.norm(v1)) # This command calculates the vector length (magnitude) of a vector named v1
```

```python
s = np.dot(v1, v2)
print(s)
```

```python
# Matrix Algebra

array1 = np.array([[1, 3], [4, 5]])
array2 = np.array([[1, 3], [4, 5]])
print(array1.size)
print(array2.size)
```

```python
array3 = array1 - array2
print(array3.size)
```

```python
m = np.array([[1,2, 3, 4, 5]])
print(m.shape)
```

```python
n = np.array([[1], [2], [3], [4] ])
print(n.shape)
```

```python
k = m * n
print(k)
```
