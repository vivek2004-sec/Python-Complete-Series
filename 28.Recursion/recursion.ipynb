```python
def mul(a, b):
    if b == 0:
        return a 
    else:
        return a + mul(a, b-1)
    
    
print(mul(5, 5))
# recursion is very much similar to stack.
```

```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n  * factorial(n-1)
    
print(factorial(5))
```

```python
def sum(n):
    if n == 0:
        return 0
    else:
        return n + sum(n - 1)
    
print(sum(7))
```
